```python
def assign_plan_type_revised(total_demand_seasonality, thresholds, Media):
    """Assigns a media plan based SOLELY on TotalDemandSeasonality and thresholds."""

    if total_demand_seasonality > thresholds[thresholds['Plan'] == 'High Intensity'][Media].iloc[0]:
        return 'High Intensity'
    elif total_demand_seasonality > thresholds[thresholds['Plan'] == 'Medium Intensity'][Media].iloc[0]:
        return 'Medium Intensity'
    elif total_demand_seasonality > thresholds[thresholds['Plan'] == 'Low Intensity'][Media].iloc[0]:
        return 'Low Intensity'
    elif total_demand_seasonality > thresholds[thresholds['Plan'] == 'Normal Period'][Media].iloc[0]:
        return 'Normal Period'
    else:
        return 'NoSpend_Period'

df['Type_of_Plan_TV'] = df.apply(
    lambda row: assign_plan_type_revised(
        row['TotalDemandSeasonality(intentionaly)'],
        MediaPlan_threshold_df,
        'TV'
    ),
    axis=1
)

df['Type_of_Plan_Digital'] = df.apply(
    lambda row: assign_plan_type_revised(
        row['TotalDemandSeasonality(intentionaly)'],
        MediaPlan_threshold_df,
        'Digital'
    ),
    axis=1
)

df['Type_of_Plan_Flyer'] = df.apply(
    lambda row: assign_plan_type_revised(
        row['TotalDemandSeasonality(intentionaly)'],
        MediaPlan_threshold_df,
        'Flyer'
    ),
    axis=1
)
```

```python
def get_media_plan_value_revised(row, media_plan_df, media):
    week = pd.to_datetime(row['Week'])
    year = week.year

    # Adjust the year for the fiscal year starting on April 1st.
    if week.month < 4:
        plan_year = year - 1
    else:
        plan_year = year

    plan_type_column = f'Type_of_Plan_{media}'
    seasonality_plan_type = row[plan_type_column]

    if seasonality_plan_type == 'NoSpend_Period':
        seasonality_value = 0
    else:
        seasonality_value = media_plan_df[
            (media_plan_df['Year'] == plan_year) &
            (media_plan_df['Media'] == media)
        ][seasonality_plan_type].iloc[0]

    # In the case of Holidays
    if row['Holidays'] == 1:
        holiday_value = media_plan_df[
            (media_plan_df['Year'] == plan_year) &
            (media_plan_df['Media'] == media)
        ]['Holidays'].iloc[0]

        # The final value will be the higher of the seasonality cost and the holiday cost.
        return max(seasonality_value, holiday_value)

    else:
        return seasonality_value

df['Plan_TV'] = df.apply(lambda row: get_media_plan_value_revised(row, MediaPlan_df, 'TV'), axis=1)
df['Plan_Digital'] = df.apply(lambda row: get_media_plan_value_revised(row, MediaPlan_df, 'Digital'), axis=1)
df['Plan_Flyer'] = df.apply(lambda row: get_media_plan_value_revised(row, MediaPlan_df, 'Flyer'), axis=1)
```

```python
#Young and Elder
# Define the lists of media types and segments to process
media_types = ['TV', 'Digital', 'Flyer']
segments = ['Young', 'Elder']

# Outer loop to iterate through each media type
for media in media_types:
    # Inner loop to iterate through each segment
    for segment in segments:
        # Dynamically generate column names using f-strings from the media and segment names
        mu_sd_column = f'Impression_{media}_{segment}'
        population_column = f'Pop_{segment}'
        base_impression_column = f'Impression_{media}'
        new_impression_column = f'Impression_{media}_{segment}'

        # Get mu and sd values
        mu = mu_sd_df[mu_sd_df['Metric'] == 'mu'][mu_sd_column].iloc[0]
        sd = mu_sd_df[mu_sd_df['Metric'] == 'sd'][mu_sd_column].iloc[0]

        # Calculate the new column and add it to the DataFrame
        df[new_impression_column] = (df[base_impression_column] *
                                     df[population_column] /
                                     df['Population'] *
                                     (norm.ppf(np.random.rand(len(df))) * sd + mu))

#Middle
df['Impression_TV_Middle'] = df['Impression_TV'] - df['Impression_TV_Young'] - df['Impression_TV_Elder']
df['Impression_Digital_Middle'] = df['Impression_Digital'] - df['Impression_Digital_Young'] - df['Impression_Digital_Elder']
df['Impression_Flyer_Middle'] = df['Impression_Flyer'] - df['Impression_Flyer_Young'] - df['Impression_Flyer_Elder']


#Column reordering
columns = df.columns.tolist()
columns = columns[:-9]
new_order = [
    'Impression_TV_Young',
    'Impression_TV_Middle',
    'Impression_TV_Elder',
    'Impression_Digital_Young',
    'Impression_Digital_Middle',
    'Impression_Digital_Elder',
    'Impression_Flyer_Young',
    'Impression_Flyer_Middle',
    'Impression_Flyer_Elder'
]
columns.extend(new_order)
df = df[columns]
```

```python
# Define the lists of media types and segments to process
media_types = ['Flyer']
segments = ['Young', 'Middle', 'Elder']

geo_column = 'Geo'

# Outer loop to iterate through each media type
for media in media_types:
    # Inner loop to iterate through each segment
    for segment in segments:
        # Dynamically generate column names
        base_reach_column = f'Reach_{media}_{segment}'
        decay_rate_L1 = eval(f'decay_rate_{media}_{segment}_L1')
        decay_rate_L2 = eval(f'decay_rate_{media}_{segment}_L2')
        decay_rate_L3 = eval(f'decay_rate_{media}_{segment}_L3')
        new_reach_L1_column = f'Reach_{media}_{segment}_L1'
        new_reach_L2_column = f'Reach_{media}_{segment}_L2'
        new_reach_L3_column = f'Reach_{media}_{segment}_L3'

        # L1: For each geo, set the new_reach_L1_column to 0 for the first week.
        df[new_reach_L1_column] = df.groupby(geo_column)[base_reach_column].transform(lambda x: x.shift(1).fillna(0)) * decay_rate_L1

        # L2: For each geo, set the new_reach_L2_column to 0 for the first to 2nd week.
        df[new_reach_L2_column] = df.groupby(geo_column)[new_reach_L1_column].transform(lambda x: x.shift(1).fillna(0)) * decay_rate_L2

        # L3: For each geo, set the new_reach_L2_column to 0 for the first to 3rd week.
        df[new_reach_L3_column] = df.groupby(geo_column)[new_reach_L2_column].transform(lambda x: x.shift(1).fillna(0)) * decay_rate_L3

#Column reordering
columns = df.columns.tolist()
columns = columns[:-9]
new_order = [
    'Reach_Flyer_Young_L1',
    'Reach_Flyer_Middle_L1',
    'Reach_Flyer_Elder_L1',
    'Reach_Flyer_Young_L2',
    'Reach_Flyer_Middle_L2',
    'Reach_Flyer_Elder_L2',
    'Reach_Flyer_Young_L3',
    'Reach_Flyer_Middle_L3',
    'Reach_Flyer_Elder_L3'
]
columns.extend(new_order)
df = df[columns]
```

```python
#Coupon Used viaDigital
# Define the lists of media types and segments to process
segments = ['Young', 'Middle', 'Elder']

# Loop to iterate through each segment
for segment in segments:
    # Dynamically generate column names using f-strings from the media and segment names
    Reach_column = f'Reach_Digital_{segment}'
    CouponRedemptionRate_column = f'CouponRedemptionRate_Digital_{segment}'
    UnitSales = eval(f'UnitSales_Coupon_viaDigital_{segment}')
    CouponDiscountRate = eval(f'CouponDiscountRate_viaDigital_{segment}')

    new_CouponUsed_column = f'CouponUsed_viaDigital_{segment}'
    new_CouponSales_column =f'CouponSales_viaDigital_{segment}'
    new_CouponDiscount_column =f'CouponDiscount_viaDigital_{segment}'

    # Get mu and sd values
    mu = mu_sd_df[mu_sd_df['Metric'] == 'mu'][new_CouponUsed_column].iloc[0]
    sd = mu_sd_df[mu_sd_df['Metric'] == 'sd'][new_CouponUsed_column].iloc[0]

    # Calculate the new column and add it to the DataFrame
    df[new_CouponUsed_column] = df[Reach_column] * df[CouponRedemptionRate_column] * df['TotalDemand'] * (1+ df['DemandShock']) *(norm.ppf(np.random.rand(len(df))) * sd + mu)
    df[new_CouponSales_column] = df[new_CouponUsed_column] * UnitSales * (1- CouponDiscountRate)
    df[new_CouponDiscount_column] = df[new_CouponUsed_column] * UnitSales * CouponDiscountRate
```

```python
#Coupon Used viaFlyer
# Define the lists of media types and segments to process
path_types = ['Directly', 'viaTV', 'viaDigital']
segments = ['Young', 'Middle', 'Elder']

# Outer loop to iterate through each path type
for path in path_types:
    # Inner loop to iterate through each segment
    for segment in segments:
        # Dynamically generate column names using f-strings from the media and segment names
        View_column = f'View_Flyer_{segment}_{path}'
        CouponRedemptionRate_column = f'CouponRedemptionRate_Flyer_{segment}'
        UnitSales = eval(f'UnitSales_Coupon_viaFlyer_{segment}')
        CouponDiscountRate = eval(f'CouponDiscountRate_viaFlyer_{segment}')

        new_CouponUsed_column = f'CouponUsed_viaFlyer_{path}_{segment}'
        new_CouponSales_column =f'CouponSales_viaFlyer_{path}_{segment}'
        new_CouponDiscount_column =f'CouponDiscount_viaFlyer_{path}_{segment}'

        # Get mu and sd values
        mu = mu_sd_df[mu_sd_df['Metric'] == 'mu'][new_CouponUsed_column].iloc[0]
        sd = mu_sd_df[mu_sd_df['Metric'] == 'sd'][new_CouponUsed_column].iloc[0]

        # Calculate the new column and add it to the DataFrame
        df[new_CouponUsed_column] = df[View_column] * df[CouponRedemptionRate_column] * df['TotalDemand'] * (1+ df['DemandShock']) *(norm.ppf(np.random.rand(len(df))) * sd + mu)
        df[new_CouponSales_column] = df[new_CouponUsed_column] * UnitSales * (1- CouponDiscountRate)
        df[new_CouponDiscount_column] = df[new_CouponUsed_column] * UnitSales * CouponDiscountRate
```
